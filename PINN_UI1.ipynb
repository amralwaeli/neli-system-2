# PINN UI1 refactor
This notebook now uses the shared engine in `engine.py` and the Gradio UI in `app.py`.
The default demo is the 1D heat equation with α = 0.1, u0 = sin(πx), and zero boundaries.

```python
from engine import PINNConfig, heat_equation_problem, solve_heat_problem

problem = heat_equation_problem(alpha=0.1)
config = PINNConfig(epochs=40, hidden_layers=2, neurons=16, n_collocation=64, n_initial=24, n_boundary=24, batch_size=1024, seed=5)
result = solve_heat_problem(problem, config, dx=0.1, dt=0.01)
print(f'Relative L2 error: {result.relative_l2_error:.6f}')
print(f'Stability number: {result.stability_number:.4f} (stable={result.stable})')
```
